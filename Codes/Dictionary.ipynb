```python
# Accessing item

my_dict = {'name':'Pooja', 'age':32, 'gender':'female'}
print(my_dict['name'])
print(my_dict['gender'])


s = {
    'name': 'Pooja',
    'College': 'VP',
    'Subjects':{
        'dsa': 40,
        'python': 60
    }
}

print(s['Subjects'])
print(s['Subjects']['python'])

# get()
print(my_dict.get('name'))
print(my_dict.get('age'))
```

```python
# Adding key value pair

info = {'name':'Pooja', 'age':32}
info['gender']='female'
print(info)


s = {
    'name': 'Pooja',
    'College': 'VP',
    'Subjects':{
        'dsa': 40,
        'python': 60
    }
}

print(s)
s['Subjects']['java'] = 76         # added new key value pair
print(s)
```

```python
# key() and values()
info = {'name':'Pooja', 'age':32, 'gender':'female'}

x = info.values()
y = info.keys()
print(x)
print(y)

# items()
x = info.items()
print(x)

# update()
info.update({'age':40})
print(info)
```

```python
# Remove key value pair
# pop() - removes item with the given keys
my_dict = {'name':'Pooja', 'age':36, 'gender':'female'}
print(my_dict)
my_dict.pop('gender')
print(my_dict)

# popitem() - removes last inserted item
my_dict = {'name':'Pooja', 'age':36,'weight':48}
print(my_dict)
my_dict.popitem()
print(my_dict)

# del - deletes item with the given keys
del my_dict['age']
#print(my_dict)

#del my_dict             # delete the complete dictionary
#print (my_dict)

my_dict.clear()
print(my_dict)
```

```python
# copy()
# returns a copy of the dictionary
my_dict = {'name':'Pooja', 'age':36, 'gender':'female'}
print(my_dict)
x = my_dict.copy()
print(x)

# any changes made in dict1 will automatically also be made in dict2,
# so we can use the built-in dict()
# mydict1 = my_dict
# print(mydict1)

mydict1 = dict(my_dict)
print(mydict1)
```

```python
# Dictinoray operations
# Membership
s = {
    'name': 'Pooja',
    'College': 'VP',
    'Sem' : 4,
    'Subjects':{
        'dsa': 40,
        'math': 60,
        'english': 55
    }
}

print('Pooja' in s)
print('name' in s)

```

```python
# Loop through a dictionary

info = {'name':'Pooja', 'age':32, 'gender':'female'}

# return value as the keys
for i in info:
  print(i)

# return values using keys()
for i in info.keys():
  print(i)
print('*' * 30)
# return value as the values
for i in info:
  print(info[i])

# return values using values()
for i in info.values():
  print(i)
print('*' * 30)
# return key,value using item()
for i in info:
  print(i,info[i])

for i,j in info.items():
  print(i,j)
```
