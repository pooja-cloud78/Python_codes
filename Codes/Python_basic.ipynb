```python
# print function
print('Hello world !!')
print("Hello world !!")

print("Hello ", 'hii')
print('Hello world !!', 77)      # string and text in one print

print(""" hi this
is python code""")

# print output single statement
""" multiline
comment"""

#Print using end and sep parameters
print ("heyy", end = " -> ")         # end parameter
print ("I will print on same line")
print('Hey', 6,7, sep = '|')        # separator parameter
```

```python
# Variables

# Printing with variable value
x = "fantastic"
print("Python is " + x)       # we can print like this also
print("Python is ", x)
```

```python
#Stylish Declaration
a=1
b=2
c=3
print(a,b,c)  # Declared separately

a=b=c=5  # Declared same value to many variables
print(a,b,c)

a,b,c=1,2,3    # Declared multiple variables at once
print(a,b,c)
```

```python
# Type Conversion
#Implicit converion
print(3+3.5)         # Automatically done by interpreter
print (type(3),type(3.5))

#Explicit conversion
#print(3+'3')
print (type(3),type('3'))
print(3+int('3'))    # Manual converion
```

```python
# Deleting a variable
x = 10
del x
print(x)  # cannot access
```

```python
# Static Vs Dynamic Binding
x = 10                  # x is int type variable
print (x)
print(type(x))

x = "Pooja"            # now x is str type variable
print (x)
print(type(x))
```

```python
# Keywords in Python
import keyword
print("The list of keywords are : ")
print(keyword.kwlist)
```

```python
#Operators
# we can do math operations in print using operators

# Arithmatic operator
print(8+5)
print(8*5)
print(8/5)
print(8//5)   # floor division
print(8%5)
print(2**3)  # 2^3 power

#Comparison operator
print(10 > 9)         #this will give result as true or false
print(10 == 9)
print(10 < 9)

#Boolean operator
print(True)
print(False)

#Logical operator
print(2 & 3)
print(2 | 3)
print(not 2)
```

```python
# Integer numbers


a=int('20',8)    # Octal to int
print (a, type(a))

a=oct(71)         # decimal to octal
print (a, type(a))

a=int('0X1e', 16)     # hexadecimal to int
print (a, type(a))
```

```python
# Floating point numbers

a=1E10             # floating point number
print (a, type(a))

# Overflow and underflow
a=1.00E400
print (a, type(a))      # Overflow

a=float("Infinity")
print (a, type(a))      # underflow
```
