```python
# Right Angled triangle

print("Right Angled triangle:")
num = 5
for i in range(1,num+1):
    print("* " * i)
```

```python
# Right Angled triangle downword

print("Right Angled triangle downword:")
num = 5
for i in range(num, 0, -1):
    print("* " * i)
```

```python
# Right Angled triangle number pattern

print("Right Angled triangle number pattern:")
num = 5
for i in range(1, num+1):
    for j in range(1,i+1):
        print(j, end=" ")
    print()
```

```python
# Right Angled triangle number pattern downword
print("Right Angled triangle number pattern downword:")
num = 5
for i in range(num, 0, -1):
    for j in range(1,i+1):
        print(j, end=" ")
    print()
```

```python
# Diamond pattern

num = 5
for i in range(1,num+1):
    for j in range(num - i):
        print(" ", end=" ")
    for k in range(2*i-1):
        print("*", end=" ")
    print()

for i in range(num-1, 0 , -1):
    for j in range(num - i):
        print(" ", end=" ")
    for k in range(2*i-1):
        print("*", end=" ")
    print()
```

```python
# hollow square
num = 5
for i in range(num):
    for j in range(num):
        if i == 0 or i == num-1 or j == 0 or j == num-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")    
    print()        
```

```python
# Print Pascal's Triangle in Python

num = 5
for i in range(num):
    print(" " * (num - i), end = " ")

    coef = 1
    for j in range(0, i + 1):
        print(coef, end = " ")
        coef = coef * (i - j)//(j + 1)
    print()    
```
