```python
# Write a function that takes a string and returns the first character that does not repeat
def first_unique(s):
    count={}
    for ch in s:
        if ch in count:
            count[ch] += 1
        else:
            count[ch] = 1

    for ch in count:
        if count[ch] == 1:
            return ch

print(first_unique("swiss"))
```

```python
# Split values greater than 1000
# Every value above 1000 should be represented by one or more chunks of 1000 plus a final remainder. If any values at or below 1000
# remain unchanged

ACC = [['A',1000],['B',900],['C',1200]]
value = 1000
output = []

for label,amount in ACC:
    while amount > value:
        output.append([label,value])
        amount -= value

    if amount > 0:
        output.append([label,amount]) 

print(output)      
```

```python
# Print all prime numbers from 1 to 100

for num in range(2,101):
    is_prime = True

    for i in range(2, int(num ** 0.5)+1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
            print(num,end=" ")
```

```python
# Find common elements between two lists

list1 = [1,2,3,4,5]
list2 = [3,4,5,6,1]

# Convert in set and used inbuilt intersection(&) function
set1 = set(list1)
set2 = set(list2)

output = set1.intersection(set2)
print(output)

# using list comprehension
result = [x for x in set1 if x in set2]
print(result)
```

```python
def sub_decorator(func):
    def wrapper(num1, num2):
        if num1 < num2:
            num1,num2 = num2, num1
        return func(num1,num2)     
    return wrapper


@sub_decorator
def substract(num1, num2):
    result = num1 - num2
    print("output is:", result)    


print(substract(4,2))
print(substract(2,4))
```

```python
# Fibonacci series
num = 10
a = 0
b = 1

# using for loop
for i in range(num):
    print(a, end=" ")
    a,b = b, a+b

# using recursion
def fibonacci(n):
    if n<=1:
        return n
    else:
        return fibonacci(n-1) + (fibonacci(n-2))

terms = 10
for i in range(terms):
    print(fibonacci(i), end = " ")    
```
