```python
# find the first non-repeating character in string using dictionaries
def non_repeat(st):
    d ={}
    for x in st:
        # d[x] = d.get(x,0) + 1
        if x in d:
            d[x] += 1 
        else:
            d[x] = 1

    for ch in st:
        if d[ch] == 1:
            return ch

str1 = "aaabbbbcdddde"
print(non_repeat(str1))  
```

```python
# Using decorator
def my_count(func):
    def wrapper(s):
        count = {}
        for i in s:
            if i in count:
                count[i] += 1
            else:
                count[i] = 1
        return count
    return wrapper
         

@my_count
def my_fun(s):
    return s

s = "aaa bbbb ccccc"
print(my_fun(s))
```

```python
# Print how many times the decorator function calls
def fun_calls(func):
    calls = 0
    def wrapper(*args, **kwargs):
        nonlocal calls
        calls += 1
        print(f"function call: {calls}")
        return func(*args, **kwargs)
    return wrapper  

@fun_calls
def greet(name):
    print(f"hello, {name}")    

greet("priya")   
greet("neha")     
```

```python
# Payment gateway and other payment options using oop concepts
# UPI payment
# Credit card
# Net banking 
from abc import ABC, abstractmethod

@abstractmethod
class payment_gateway(ABC):
    def pay(self):
        raise NotImplementedError("subclasses must implement the method")


class Upi_payment(payment_gateway):
    def pay(self,amount):
        print(f"Payment of rs {amount} processed using UPI")
        
class CreditCard(payment_gateway):
    def pay(self,amount):
        print(f"Payment of rs {amount} processed using credit card")  

class NetBanking(payment_gateway):
    def pay(self,amount):
        print(f"Payment of rs {amount} processed using NetBanking")         

obj = [Upi_payment(),CreditCard(),NetBanking()]    

for i in obj:
    i.pay(1000)
```

```python
def handledivisionerror(func):
    def wrapper(a,b):
        if b == 0:
            return "Cannot divide by zero"
        return func(a,b)
    return wrapper

@handledivisionerror
def division(a,b):
    return a / b

print(division(10,0))
```

```python
#Keep the latest record and remove the duplicate record
records = [
    {"cid": 101, "updated": "2026-10-20"},
    {"cid": 102, "updated": "2026-10-21"},
    {"cid": 101, "updated": "2026-10-24"},
    {"cid": 103, "updated": "2026-10-22"}
]
new_record = {}

for rec in records:
    cust_id = rec["cid"]

    if (cust_id not in new_record or rec["updated"] > new_record[cust_id]["updated"]):
        new_record[cust_id] = rec

print(new_record.values())    
```

```python
# Find highest salary in each department
employees = [
    {"name":"amit","salary": 50000, "dept": "IT"},
    {"name":"priya","salary": 70000, "dept": "HR"},
    {"name":"rahul","salary": 65000, "dept": "IT"},
    {"name":"sneha","salary": 80000, "dept": "HR"}
]

IT_list = {}
HR_list = {}

for emp in employees:
    dep = emp["dept"]
    if dep == "IT":
        if dep not in IT_list or IT_list["salary"] > emp["salary"]:
            IT_list = emp
    else:
        if dep not in HR_list or HR_list["salary"] > emp["salary"]:
            HR_list = emp

print("Highest salary in IT:",IT_list)
print("Highest salary in HR:",HR_list)
```
