```python
# Parameter vs Argument

# Types of Argument
  # Dafault Argument
  # Positional Argument
  # Keyword Argument
```

```python
# Dafault Argument

def my_function(name = "friend"):
  print("Hello", name)

my_function("Pooja")
my_function()
```

```python
# Keyword arguments(kwargs)
# values send in the key=value syntax

def my_function(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

# order of arguments doesnt matter
my_function(animal = "dog", name = "Buddy")
my_function(name = "Buddy",animal = "dog")

```

```python
# Positional arguments

def my_function(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

my_function("dog", "Buddy")     # arguments must be in correct order
```

```python
# Mixing positional and keyword arguments

def my_function(animal, name, age):
  print("I have a", age, "year old", animal, "named", name)

my_function("dog", name = "Buddy", age = 5)   # positional arg must come before keyword arg
```

```python
# Passing different data types
# list as an argument

def my_function(fruits):
    print(fruits)

my_fruits = ["apple", "banana", "cherry"]
my_function(my_fruits)

# dictionary as an argument

def my_function(person):
  print("Name:", person["name"])
  print("Age:", person["age"])

my_person = {"name": "Pooja", "age": 25}
my_function(my_person)
```

```python
# Function returns different data types
# returns a list from function

def my_function():
  return ["apple","banana","cherry"]

fruits = my_function()
print(fruits[0])
print(fruits[1])

# returns a list from function

def my_function():
  return (10, 20)

x, y = my_function()
print("x:", x)
print("y:", y)
```

```python
# Positional only arguments
def my_function(name, /):
  print("Hello", name)

my_function("Pooja")    # cant use keyword arguments here

# keyword only arguments
def my_function(*, name):
  print("Hello", name)

my_function(name = "Pooja")      # cant use positional arguments here

# Combining positional only and keyword only arguments
def my_function(a, b, /, *, c, d):
  return a + b + c + d

result = my_function(5, 10, c = 15, d = 20)
print(result)
```

```python
# Arbitrary arguments- *args

def my_function(*kids):                  # receive a tuple of arguments
  print("The youngest child is " + kids[2])

my_function("Pooja", "Teja", "Prafulla")


# Arbitrary keyword arguments- **kwargs
def my_function(**kid):
  print("Her last name is " + kid["lname"])

my_function(fname = "Pooja", lname = "Dhere")

# Combining *args and **kwargs
def my_function(title, *args, **kwargs):        # This should be order(regular,*args,**kwargs)
  print("Title:", title)
  print("Positional arguments:", args)
  print("Keyword arguments:", kwargs)

my_function("User Info", "Pooja", "Teja", age = 25, city = "Pune")
```

```python
# Unpacking arguments
# Unpacking list using *
def my_function(a, b, c):
  return a + b + c

numbers = [1, 2, 3]
result = my_function(*numbers) # Same as: my_function(1, 2, 3)
print(result)

# Unpacking dictionary using **

def my_function(fname, lname):
  print("Hello", fname, lname)

person = {"fname": "Pooja", "lname": "Dhere"}
my_function(**person)
```

```python
# Scope of variables
# local scope
def myfunc():
  a = 800        # local variable
  print(a)

myfunc()
#print(a)         # cannot access outside of local scope
```

```python
# global scope
y = 300          # global variable
def myfunc():
  print(y)

myfunc()
print(y)

# if local and global variable shares the same name, python will treat them different
x1 = 700
def myfunc():
  x1 = 100
  print(x1)

myfunc()
print(x1)

# with global keyword we can make variable as global even inside function also
z = 280
def myfunc():
  global z
  z = 400

myfunc()
print(z)
```

```python
# Enclosing scope (nonlocal)
# In nested functions variables can be accessed from outer function to inner function using nonlocal keyword

def myfunc1():
  x = "Jane"
  def myfunc2():
    nonlocal x
    x = "hello"

  myfunc2()
  return x

print(myfunc1())
```

```python
# LEGB rule

# Local - Inside the current function
# Enclosing - Inside enclosing functions (from inner to outer)
# Global - At the top level of the module
# Built-in - In Python's built-in namespace

x = "global"

def outer():
  x = "enclosing"
  def inner():
    x = "local"
    print("Inner:", x)
  inner()
  print("Outer:", x)

outer()
print("Global:", x)
```

```python
# Decorators
# let you add extra behaviour to the function without changing the original code

def changecase(func):                 # decorator
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():             # function that gets decorated
  return "Hello Pooja"

print(myfunction())
```

```python
# Arguments in the decorated function

def changecase(func):
  def inner(x):
    return func(x).upper()
  return inner

@changecase
def my_func(name):
   return "Hello " + name

my_func("Pooja")
```

```python
# Any number of arguments and any type of arguments can be passed to the wrapper function
# Add *args, **kwargs to the wrapper function

def changecase(func):
  def inner(*args,**kwargs):
    return func(*args,**kwargs).upper()
  return inner

@changecase
def my_func(name):
   return "Hello " + name

my_func("Pooja")
```

```python
# Decorator with arguments
def changecase(n):
  def changecase(func):
    def myinner():
      if n == 1:
        a = func().lower()
      else:
        a = func().upper()
      return a
    return myinner
  return changecase

@changecase(1)
def myfunction():
  return "Hello Python"

print(myfunction())
```

```python
# Preserving function metadata (__name__)
# using functools.wraps

import functools
def changecase(func):
  @functools.wraps(func)
  def myinner():
    return func().upper()
  return myinner

@changecase
def my_function():
  return "hello"

print(my_function())
print(my_function.__name__)    # returns function name
```

```python
# Lambda functions

# lambda function takes one argument
x = lambda a : a + 10
print(x(5))

# lambda function takes two argument
x = lambda a,b : a * b
print(x(2,3))

# lambda function takes three argument
x = lambda a,b,c: a + b + c
print(x(1,8,9))
```

```python
# Lambda function in another function
def myfunc(n):
  return lambda a: a * n

mydoubler = myfunc(2)
print(mydoubler(11))
print(mydoubler(10))
print(mydoubler(111))

mytripler = myfunc(3)
print(mytripler(11))
print(mytripler(111))
```

```python
# Higher order functions
# 1) Passing a function as an argument
def apply(func, x):
    return func(x)

def square(n):
    return n * n

print(apply(square, 5))  # 25

# 2) Returning a function
def multiplier(n):
    return lambda x: x * n

double = multiplier(2)
print(double(5))  # 10
```

```python
# 3) Built-in function
# map()
list(map(lambda x: x**2, [1, 2, 3, 4]))

# another way to write same
numbers = [1,2,3,4]
doubled = list(map(lambda x: x**2,numbers))
print(doubled)

# filter()
list(filter(lambda x: x%2 == 0,[1,2,3,4,5,6]))

def even(n):
    return n % 2 == 0

a = [1, 2, 3, 4, 5, 6]
b = filter(even, a)
print(list(b)) # Convert filter object to a list

# filter function checks for all items and removes falsy values
L = ["apple", "", None, "banana", 0, "cherry"]
A = filter(None, L)
print(list(A))

# sorted() with key =
words = ["python", "a", "function", "code"]
sorted(words, key=len)
```

```python
# Function Recursion
# Function calls itself
def my_function(num):
  if num <= 0:
    print("Done!")
  else:
    print(num)
    my_function(num - 1)

my_function(5)


# Factorial using recursion

def fact(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n * fact(n - 1)

print(fact(5))
```

```python
# Recursion with list

def sum_list(my_list):
  if len(my_list) == 0:
    return 0
  else:
    return my_list[0] + sum_list(my_list[1:])

numbers = [1,2,3,4,5]
result = sum_list(numbers)
print(result)
```

```python
# Generators
# functions which can pause and resume their execution

def my_generator():
  yield 1
  yield 2
  yield 3

for value in my_generator():
  print(value)
```

```python
# Generators saves memory
def large_sequence(n):
  for i in range(n):
    yield i

gen = large_sequence(100000)
print(next(gen))
print(next(gen))
print(next(gen))

# StopIteration exception
def simple_gen():
  yield 1
  yield 2

gen = simple_gen()
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# Generator expression
# It creates a generator

gen_exp = (x * x for x in range(5))
print(gen_exp)
print(list(gen_exp))


# sum of elements
gen_exp= sum(x * x for x in range(5))
print(gen_exp)
```

```python
# Generators method
# send() method
def my_gen():
  while True:
    received = yield
    print("received:",received)

gen = my_gen()
next(gen)
gen.send("Hello")
gen.send("World")


# close() method
def echo_generator():
  try:
    yield 1
    yield 2
    yield 3
  finally:
    print("Generator closed")

gen = echo_generator()
print(next(gen))
gen.close()
```
