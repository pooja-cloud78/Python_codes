```python
# String: reverse, count vowels, check palindrome, count words, change case

def reverse_string(name):
    print("reverse string: ",name[::-1])

def count_vowels(name):
    vowels = "aeiouAEIOU"
    count = 0
    for i in name:
        if i in vowels:
            count +=1
    print("vowels count: ",count)

def check_palindrome(name):
    rev = name[::-1]
    if name == rev:
        print("string is palindrome")
    else:
        print("string is not palindrome")

def count_words(name):
    count = len(name.split())
    print("count of words: ",count)

def change_case(name):
    print("change case: ",name.swapcase())

str1 = input("Enter string:")

while True:
  print("\nChoose an operation:")
  print("1. Reverse String")
  print("2. Count Vowels")
  print("3. Check Palindrome")
  print("4. Count Words")
  print("5. Change Case")
  print("6. Exit")

  choice = int(input("enter choice number:"))

  match choice:
      case 1:
        reverse_string(str1)
      case 2:
        count_vowels(str1)
      case 3:
        check_palindrome(str1)
      case 4:
        count_words(str1)
      case 5:
        change_case(str1)
      case 6:
        break
      case _:
        print("Enter valid operation")
```

```python
# String: slicing, split(), join(), replace(), f-string formatting
# slicing
str1 = "hello"
print(str1[1:4])

# split()
str2 = "  abcd   "
print(str2.split())

# join()
x = '#'
print(x.join(str1))

# replace()
s = "I am learning java"
s1 = s.replace('java','python')
print(s1)

# f-string format
age = 40
print(f"I am {age} years old")
```

```python
# List: find max/min/sum/average without built-ins

def max_min_list(list1):
    max_num = list1[0]
    min_num = list1[0]

    for i in range(len(list1)):
        if list1[i] > max_num:
          max_num = list1[i]
        elif list1[i] < min_num:
          min_num = list1[i]
    print("max num",max_num)
    print("min num",min_num)

def sum_avg_list(list1):
    sum_num = 0
    for i in range(len(list1)):
        sum_num += list1[i]
    print("sum",sum_num)
    avg_num = sum_num / len(list1)
    print("average", avg_num)

list1 = [1,2,3,9,5]
print(list1)
max_min_list(list1)
sum_avg_list(list1)
```
