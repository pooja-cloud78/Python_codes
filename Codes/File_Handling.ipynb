"r" - Read - Default value. Opens a file for reading, error if the file does not exist

"a" - Append - Opens a file for appending, creates the file if it does not exist

"w" - Write - Opens a file for writing, creates the file if it does not exist

"x" - Create - Creates the specified file, returns an error if the file exists

```python
# Create a new file

f = open("myfile.txt", "x")

#If the file already exist, an error will be raised.
# a new empty file is created.
```

```python
# Opening and closing a file
# If the file exists, it returns a file object connected to that file
# if the file does not exist, Python raises a FileNotFoundError.

file = open("abc.txt", "r")
print(file)         #returns a file object
print(file.read())
# Perform file operations
file.close()          # releases the system resources with closing a file
```

```python
# Opening a file using with statement
# file will close automatically

with open("abc.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
# Checking file properties
f = open("abc.txt", "r")
print("Filename:", f.name)
print("Mode:", f.mode)
print("Is Closed?", f.closed)

f.close()
print("Is Closed?", f.closed)
```

```python
# Reading a file
# If file is not exists it shows error
# read() − Reads the entire file.
file = open("abc.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
# readline() − Reads one line at a time.

with open("abc.txt", "r") as file:
    # print(file.readline()))           # returns first line only
   line = file.readline()
   while line:
      print(line, end='')
      line = file.readline()
```

```python
# readlines − Reads all lines into a list.

with open("abc.txt", "r") as file:
    # print(file.readlines())           # returns all lines into list
   lines = file.readlines()
   for line in lines:
      print(line, end='')
```

```python
# Reading a part of line
with open("abc.txt", "r") as file:
    print(file.read(5))       # reads first 5 characters
```

```python
# Writing a file
# This creates a new file if it doesn’t exist, or overwrites the existing file if it does.
# with statement, which automatically handles closing
# using write() method

with open("abc.txt", "w") as file:
    file.write("Hello, Python!\n")
    file.write("File handling is easy")

print("File written successfully")
```

```python
# using writelines() method
# To take a list of strings and writes each string to the file
# Multiple lines at once

lines = ["First line\n", "Second line\n", "Third line\n"]
with open("xyz.txt", "w") as file:
   file.writelines(lines)
   print ("Content added Successfully!!")
```

```python
# append mode
# Write in an existing file at the end of file

with open("xyz.txt", "a") as file:
    file.write("file has more content")

#open and read the file after the appending:
with open("xyz.txt") as f:
  print(f.read())
```

```python
# Delete a file
# To delete a file, you must import the OS module, and run its os.remove() function

import os
os.remove("myfile.txt")

# Check if file exists, then delete it:

import os
if os.path.exists("myfile.txt"):
  os.remove("myfile.txt")
else:
  print("The file does not exist")
```

```python
# Handling Exceptions When Closing a File

try:
    with open("abc.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError as e:
    print("Error:", e)
```
