```python
# Creating Tuple
  # empty
t1 = ()
print(t1)
print(type(t1))

# Creating tuple with single item
t2 = (2)
print(t2)
print(type(t2))

t3 = ('hello')
print(t3)
print(type(t3))

t4 = (1,2,3,4)
print(t4)
print(type(t4))

t5 = (2,)
print(t5)
print(type(t5))
```

```python
# Tuple can be created without parenthesis

thistuple = "apple", "banana", "cherry"
print(thistuple)
print(type(thistuple))

t4 = 1,2,3,4
print(t4)
print(type(t4))
```

```python
# Using list
li = [1,2,3,4,5]
print(tuple(li))

tup = tuple('hello')
print(tup)
```

```python
# heterogenous tuple
tup = (5,'hello',7,5,True,[1,2],{'key':'value'})
print(tup)

t1 = (1,3,True,(9,3.5))
print(t1)
```

```python
# Indexing
t1 = (1,2,3)
print(t1)
print(t1[0])
print(t1[-1])
#print(t1[-6])
```

```python
# Slicing

t = (1,2,3,4)
print(t[0:4])
print(t[0:4:2])
print(t[-3:-1])
print(t[::-1])    #reverse tuple


t5 = (1,2,3,(4,5))
print(t5)
print(t5[-1][0])
```

```python
# Update tuple
# Tuples are unchangeable, meaning that you cannot change, add, or remove items once the tuple is created.
t = (1,2,3,4)
print(t)
#t[0]=100
#print(t)

# Other ways to update tuples
# 1.convert into list

thistuple = ("apple", "banana", "cherry")
y = list(thistuple)
y.append("orange")
thistuple = tuple(y)
print(thistuple)

# 2. add tuple into tuple
thistuple = ("apple", "banana", "cherry")
y = ("orange",)
thistuple += y

print(thistuple)
```

```python
# Removing tuple items
# We cannot delete tuple items as they are unchangeable

t = (1,2,3,4,5)
#del t[-1]

#Convert it into list
y = list(t)
y.remove(1)
t = tuple(y)
print(t)

#delete tuple completely
del t
#print(t)
```

```python
# Operation on tuple

# + and *

t1 = (1,2,3,4)
t2 = (4,5,6,7)
print(t1+t2)
print(t1*3)

# Membership
print(1 in t1)
print(6 in t1)
print(6 not in t2)

#Looping through items
for i in t1:
  print(i)
```

```python
#Tuple Unpacking

# Packing a tuple
fruits = ("apple", "banana", "cherry")

# Unpacking a tuple
fruits = ("apple", "banana", "cherry")

(green, yellow, red) = fruits     # no. of items must match

print(green)
print(yellow)
print(red)

#swapping of elements
a = 1
b = 2
a,b = b,a
print(a,b)

# using asterisk(*)

fruits = ("apple", "mango", "papaya", "pineapple", "cherry")

(green, *tropic, red) = fruits

print(green)
print(tropic)        #collect elements in list
print(red)

a,b,*c=(1,2,3,4)
print(a,b)
print(*c)         #collect remaining element in list
```

```python
# Tuple methods
# len/min/max/sorted/sum
t = (8,7,23,6,89,45)
print(len(t))
print(min(t))
print(max(t))
print(sum(t))
print(sorted(t))

# count

print(t.count(23))

# Index
print(t.index(7))
#print(t.index(36))
```

```python
# Difference between List and Tuple
# Tuple takes less time than list

import time
L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*6
print('List Time',time.time()-start)

start = time.time()
for i in T:
  i*6
print('Tuple Time',time.time()-start)
```

```python
# Memory

import sys
L = list(range(1000))
T = tuple(range(1000))
print('List Size',sys.getsizeof(L))
print('Tuple Size',sys.getsizeof(T))
```
