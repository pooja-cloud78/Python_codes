```python
# Check for wrong email and password with second attempt

email = input('Enter your email')
password = input('Enter Your Password')

if email == 'poojadhere@gmail.com' and password == '123':
  print('Welcome',email)
elif email != 'poojadhere@gmail.com' and password == '123':
    print('Incorrect email')
    email = input('Enter your email again: ')
    if email == 'poojadhere@gmail.com':
      print('Welcome',email)
    else:
      print('Your attempt is over, Please check with your admin')
elif email == 'poojadhere@gmail.com' and password != '123':
  print('Incorrect Password')
  password = input('Enter Your Password again: ')
  if password == '123':
    print('Welcome',email)
  else:
    print('Your attempt is over, Please check with your admin')
else:
  print('Wrong inputs')
```

```python
# Write a program to find out whether a student has passed or failed if it requires a total of
# 40% and at least 33 in each subject to pass. Assume 3 subjects and take marks as an
# input from the user

math = int(input('Enter math marks: '))
science = int(input('Enter science marks: '))
history = int(input('Enter history marks: '))

total_marks = math + science + history
per_marks = (total_marks/300) * 100

if per_marks >= 40:
  if math <= 33:
    print('student has failed in math')
  elif science <= 33:
    print('student has failed in science')
  elif history <= 33:
    print('student has failed in history')
  else:
    print('student has passed')
else:
  print ('student has failed')
```

```python
# diamond pattern
rows = int(input("enter rows: "))

# upper half
for i in range(1, rows + 1):
  for j in range(rows - i):
    print(" ", end = " ")
  for k in range (2 * i - 1):
    print("*", end = " ")
  print()

# lower half
for i in range(rows - 1, 0, -1):
  for j in range(rows - i):
    print(" ", end = " ")
  for k in range (2 * i - 1):
    print("*", end = " ")
  print()
```

```python
# Print Pascal's Triangle in Python

rows = int(input("enter rows: "))
# iterate up to rows
for i in range(rows):
    # adjust space
    print(' '*(rows-i), end='')

    # compute each value in the row
    num = 1
    for j in range(0, i + 1):
        print(num, end=' ')
        num = num * (i - j) // (j + 1)
    print()
```

```python
# Hollow square
rows = 5
for i in range(rows):
    for j in range(rows):
        if i == 0 or i == rows - 1 or j == 0 or j == rows - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Write a program to find whether a given number is prime or not

n = int(input("Enter number: "))

if n <= 1:
    print(False)
else:
    prime = True

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            prime = False
            break
    print(prime)
```

```python
# Password check with multiple attempts

correct_pwd = "python123"
attempts = 3
while attempts > 0:
  password = input("Enter password: ")
  if password == correct_pwd:
    print("Correct password")
    break
  else:
    attempts -= 1
    print("Wrong password, attempts left",attempts)

if (attempts == 0):
      print("Too many failed attempts")
```

```python
# Continue statement
# stop the current iteration of the loop and continue with the next one

for i in range(4):
  print("printing")
  if i == 2:      # if i is 2, the iteration is skipped
    continue
  print(i)
```
