**String**

```python
# String representation
s = 'hello'
s = "hello"
s = '''hello'''
s = """hello"""
print(s)

s= str("hello")
print(s)

a = "it's a python learning"
print(a)

x = 'this is "python" course'
y = "this is 'python' course"
print(x)
print(y)
```

```python
# Accessing elements and Slicing
s = "hello world"
# Accessing string elements
print(s[0])
print(s[2])

print(s[2:])
print(s[2:3])
print(s[:11])
print(s[:])
print(s[0:6:2])
print(s[1:9:3])

print(s[6:0:-1])
print(s[::-1])   #reverse string

#Negative indexing
print(s[-6:0]) #returns empty space
print(s[-5:])
print(s[-8:-2])

s = "ABCDEF"
print(s[0:5:2])
print(s[-3:])
print(s[-5:-1])
```

```python
# Editing and deleting string
s = "hello world"
print(s[0])

# s[1]='P'
# Python strings are immutable

del s
#print(s)
```

```python
# operations on string
# Arithmatic operator(+ and *)
print("pune" + " " + "mumbai")
print('pune ' * 5)

print('#' * 30)
```

```python
# Relational operator
print('pune' == 'Pune')     # case sensitive
print('pune' == 'pune')
print('pune' != 'Pune')     # works with ascii value not by characters
print('pune' > 'Pune')
```

```python
# Logical operators
# works with binary values
print('hello' and 'world')
print('hello' or 'world')
```

```python
# Loop with string
for i in 'pune':
  print(i)

for i in 'pune':
  print('hi')

print('p' in 'pune')
print('p' not in 'pune')
```

```python
# f string
name = "Jake"
age = 22
print(f"Name: {name}, Age: {age}")

age = 36
txt = f"My name is John, I am {age}"
print(txt)

# modifier .2f
price = 59
txt = f"The price is {price:.2f} dollars"
print(txt)

# placeholder can contain math operation
txt = f"The price is {20 * 59} dollars"
print(txt)

s = "My name is {} and I am {} years old.".format( "Emily", 22)
print(s)
```

```python
# String functions
# length
print(len("hello"))

# min value(depending on ascii)
print(min("hello"))

# max value(depending on ascii)
print(max("hello"))

# sorted
print(sorted("hello")) #(ascending order by default) return in list format
print(sorted("hello", reverse = True))  # descending order
```

```python
# Capitalize/Title/upper/lower
s = 'hello world'
print(s.capitalize())
print(s.title())
print(s.upper())
print(s.lower())
print(s.swapcase())  #opposite of original
```

```python
# Count/Find/index/startswith/endswith
s = 'hello world'
print(s.count('o'))   # count of character
print(s.find('o'))    # find given characters position
print(s.find('k'))    # not presend in string

# Index() is similar to find() except if given character is not present in string
# it will give error where find() will return -1

print(s.index('o'))
#print(s.index('k'))   # throw error

print(s.startswith('e'))  # return false
print(s.startswith('h'))  # return true

print(s.endswith('e'))  # return false
print(s.endswith('d'))  # return true
```

```python
# replace()
s = 'hello'
print(s.replace('o','a'))  # count = -1 by default means replace all occurances
print(s.replace('l','p',count=1))  # if optional it will replace first occurance
print(s.replace('l','p',count=2))
```

```python
# join()
x = '#'
print(x.join('hello'))

# f-string format
age = 40
print(f"I am {age} years old")
```
