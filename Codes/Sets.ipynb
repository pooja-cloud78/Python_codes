```python
# Creating sets

# empty set

s = {}
print(s)
print(type(s))      # Empty set is of dict type

s1 = {1,2,3,4}
print(s1)
print(type(s1))

# created using set constructor
s5 = set([1,2,3])
print(s5)

s6 = set(("apple","banana","cherry"))
print(s6)

# Heterogenous set
s5 = {2,4.5,True,7,'pune'}
print(s5)

# Unorderd
s1 = {1,2,3}
s2 = {3,2,1}
print(s1==s2)
```

```python
# set cant contain mutable data types(set,list)
# s2 = {1,2,3,{4,5}}
# print(s2)
# s3 = {1,2,3,[4,5]}
# print(s3)

s4 = {1,2,3,(4,5)}   # tuple is immutable so it can be added
print(s4)

# Duplicate not allowed
s8 = {1,2,5,6,3,2,2,1,5,6}
print(s8)

# True and 1 treated as same and considered as duplicate
set1 = {"apple",True,4,1,2}
print(set1)

# False and 0 treated as same and considered as duplicate
set2 = {"apple",False,4,0,2}
print(set2)
```

```python
# Accessing items does not work becouse its unorderd

s1 = {1,2,3,4,5}
print(s1)
print(s1[0])

# Editing
s3 = {1,2,3}
s3[0] = 100
print(s3)

# editing, updating or slicing is not possible in sets
```

```python
# Adding set items
# add() to add items
s = {1,2,3,4}
s.add(50)
print(s)

#update() to add items from another set into current set
s1 = {1,2,3,4}
s1.update([9,8,7])
print(s1)

s2 = {1,2,3,4}
s3 = {3,4,7,8,9}
s2.update(s3)     # duplicates will not add
print(s2)

# The object in the update() method does not have to be a set
# it can be any iterable object (tuples, lists, dictionaries etc.)
```

```python
# Removing set items
# remove()
s1 = {1,2,3,4,5}
s1.remove(5)
print(s1)

# s1.remove(60)       # if item does not exists,it will raise an error

# discard()
s = {1,2,3,4,5}
s.discard(3)
print(s)

s.discard(60)        # If the item to remove does not exist, discard() will NOT raise an error.
print(s)

# pop() removes random item
s4 = {1,2,3,4,5}
print(s4)
s4.pop()
print(s4)


thisset = {"apple", "banana", "cherry"}
x = thisset.pop()
print(x)                     #removed item
print(thisset)                #the set after removal
```

```python
# Join sets

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# union (|)
s3 = s1.union(s2)      # Add items from both sets without duplicates
s4 = s1 | s2

print(s1|s2)
print(s3)
print(s4)

#Join multiple sets

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"apple", "banana", "cherry"}

myset = set1.union(set2, set3)    # using union()
print(myset)

myset1 = set1 | set2 | set3         # using | operator
print(myset1)

#Join a set with a tuple
x = {"a", "b", "c"}
y = (1, 2, 3)

z = x.union(y)       # only set with set can join using | operator not with other datatype
#z = x | y          # like we can do with union()
print(z)
```

```python
# Intersection (&)
# items present in both sets
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.intersection(set2)
print(set3)

set4 = set1 & set2
print(set4)


s1 = {"apple", 1,  "banana", 0, "cherry"}
s2 = {False, "google", 1, "apple", 2, True}

s3 = s1.intersection(s2)

print(s3)

# intersection_update()
set1 = {1,2,3,5}
set2 = {5,6,7,8}

set1.intersection_update(set2)        # Change the original set
print(set1)
```

```python
# difference() , (-)
# items present in first set but not in other set
set1 = {1,2,3,4}
set2 = {4,5,6,7}

set3 = set1.difference(set2)
print(set3)

set4 = set1 - set2
print(set4)

# difference_update() , (-=)
set1 -= set2
print(set1)

s1 = {3,4,9,5}
s2 = {5,6,7,8}

s1.difference_update(s2)        # Change the original set
print(s1)
```

```python
# symmetric_difference(), (^)
# all different elements from both sets
set1 = {1,2,3,4}
set2 = {4,5,6,7}

set3 = set1.symmetric_difference(set2)
print(set3)

set4 = set1 ^ set2
print(set4)

# symmetric_difference_update()
s1 = {1,2,3,5}
s2 = {5,6,7,8}

s1.symmetric_difference_update(s2)        # Change the original set
print(s1)
```

```python
# copy()
fruits = {"apple", "banana", "cherry"}
x = fruits.copy()           # returns a copy of a set
print(x)
```

```python
# isdisjoint
# if all items are different in set then it will return True otherwise it will return false
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1.isdisjoint(s2))
s3 = {1,2,3,4}
s4 = {7,8,9,10}
print(s3.isdisjoint(s4))

# issubset(), (<=)
s5 = {1,2,3,4,5,6}
s6 = {3,4,5}

print(s5.issubset(s6))
print(s6.issubset(s5))
print(s6 <= s5)    # using <= operator

# issuperset(), (>=)
s7 = {1,2,3,4,5,6}
s8 = {3,4,5}
print(s7.issuperset(s8))
print(s8.issuperset(s7))
print(s7 >= s8)    # using >= operator
```

```python
# Frozenset (immutable version of set)

#s = {1,2,3,[4,5]} # cant add list in set as its mutable
#print(s)

fs = frozenset([1,2,3])
print(fs)
fs1 = frozenset((1,2,3))
print(fs1)
fs2 = frozenset({1,2,3})
print(fs2)
```

```python
# Set Comprehension

print({i for i in  range(1,11)})
print({i for i in  range(1,11) if i > 5})
print({i for i in  range(1,11) if i < 5})

square_or_minus_one = {num **2 if num % 2 == 0 else -1 for num in range(1,11)}
print(square_or_minus_one)

square_or_minus_one = {num * num if num % 2 == 0 else -1 for num in range(1,11)}
print(square_or_minus_one)
```
