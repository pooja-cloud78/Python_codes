```python
# Default constructor
class person:
  def __init__(self):                # constructor
    self.name = "Pooja"          # set default values
    self.age = 32

p1 = person()
print(p1.name, p1.age)

# parameterized constructor
class person:
  def __init__(self,name,age):
    self.name = name
    self.age = age

p1 = person("Bob", 37)            # Different objects can have different value
p2 = person("karan", 40)

print(p1.name, p1.age)
print(p2.name, p2.age)

# Default values can be given in constructor
class A:
  def __init__(self,name,age=18):
    self.name = name
    self.age = age
a1 = A("Alice")
print(a1.name, a1.age)
```

```python
class A:
  def __new__(cls):                    # creating object
    print("Creating instance")
    return super().__new__(cls)

  def __init__(self):                 # initializing object
    print("Initializing instance")

A()     # object creation
```

```python
# self parameter
class person:
  def __init__(self,name,age):
    self.name = name
    self.age = age

p1 = person("karan", 29)
print(p1.name)
print(p1.age)
```

```python
# Instead of self we can use any other name also
class person:
  def __init__(abc,name,age):
    abc.name = name
    abc.age = age
    print(abc)        # reference to the current object

p1 = person("karan", 29)
print(p1)              # same address as self(abc)
print(p1.name)
print(p1.age)

# self is a parameter which reference to the object
```

```python
# Instance(object) properties
class Person:
  def __init__(self, name, age):
    self.name = name     # Instance variable
    self.age = age       # Instance variable

p1 = Person("Pooja", 25)
print(p1.name)           # access instance variable

p1.age = 30           # modify instance variable outside class
print(p1.age)

p1.gender = "female"       # create properties outside class
print(p1.gender)          #Adding properties this way only adds them to that specific object, not to all objects of the class.

del p1.gender             # delete class properties
# print(p1.gender)
```

```python
# class properties

class Person:
  lastname = ""         # class variable

  def __init__(self, name):
    self.name = name         # instance variable

p1 = Person("Pooja")

Person.lastname = "Dhere"      # Modify class property

print(p1.name)
# print(Person.name)       # instance variable cant be accessed with class name

print(p1.lastname)         # class variables can be accessed with objects also
print(Person.lastname)

Person.city = "Pune"            # Add new property
print(p1.city)
```

```python
# class method
class Person:
  def __init__(self, name, age):      # constructor
    self.name = name
    self.age = age

  def get_info(self):        # class method
    return f"{self.name} is {self.age} years old"

p1 = Person("Pooja", 30)
print(p1.get_info())
```

```python
# Class method
class Calculator:
  def add(self, a, b):     # class method
    return a + b

  def multiply(self, a, b):   # class method
    return a * b

calc = Calculator()
print(calc.add(5, 3))
print(calc.multiply(4, 7))
```

```python
# Pass by reference
# We can pass class object to function as a argument

class Person:
  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

def greet(Person):
    print(Person.name, Person.gender)

p = Person('Rahul','male')
greet(p)
```

```python
# Object Mutability
# Object of user defined class is always mutable just like list,set,dict

class Person:
  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside class

def greet(person):
  person.name = 'Ankita'

p = Person('Rahul','male')
greet(p)
print(p.name)
```

```python
# static method
# methods that dont use the self parameter
class student:
  @staticmethod
  def add(a, b):
    print("Addition:",a + b)

student.add(5,3)    # call via class

s = student()  # call via instance
s.add(2,7)
```

```python
# Encapsulation
class Person:
  def __init__(self, name, age):
    self.name = name
    self.__age = age # Private property

p1 = Person("Emil", 25)
print(p1.name)
p1.__age = 30   # We can change the value but we cannot access it
# print(p1.__age) # This will cause an error
```

```python
# Private methods or attributes can be accessed only inside class
class account:
  def __init__(self, acc_no, acc_pass):
    self.acc_no = acc_no
    self.__acc_pass = acc_pass

  def reset_pass(self):
    print(self.__acc_pass)

acc = account("123456","abc123")
print(acc.acc_no)
print(acc.reset_pass())
```

```python
#We can access private methods not directly but through other public methods outside the class
class person:
  __name = "pooja"

  def __hello(self):
    return self.__name

  def welcome(self):
    print(self.__hello())

p1 = person()
p1.welcome()
```

```python
# Inheritance
# We can inherit the properties and methods from another class(parent)
# single inheritance
class car:
  @staticmethod
  def start():
    print("car started")

  @staticmethod
  def stop():
    print("car stopped")

class toyota(car):
  def __init__(self,type):
    self.type = type

t = toyota("petrol")
t.start()
t.stop()
```

```python
# Multilevel inheritance

class car:
  @staticmethod
  def start():
    print("car started")

  @staticmethod
  def stop():
    print("car stopped")

class toyota(car):
  def __init__(self, brand):
    self.brand = brand

class fortuner(toyota):
  def __init__(self,type):
    self.type = type

car1 = fortuner("diesel")
car1.start()
car1.stop()
```

```python
# Multiple inheritance
class A:
  var1 = "class_A"

class B:
  var2 = "class_B"

class C(B,A):
  var3 = "class_C"

c1 = C()
print(c1.var1)
print(c1.var2)
print(c1.var3)
```

```python
class Employee:
    def __init__(self, name, age):
        self.name = name       # public
        self._age = age        # protected

class SubEmployee(Employee):
    def show_age(self):
        print("Age:", self._age)   # Accessible in subclass

emp = SubEmployee("Ross", 30)
print(emp.name)        # Public accessible
print(emp._age)        # Can access protected members but we shoudn't
emp._age = 40
emp.show_age()         # Protected accessed through subclass
```

```python
class Animal:
  def __init__(self, name):
    self.name = name
    print("Animal constructor", self.name)

  def info(self):
    print("Animal name:", self.name)

class Dog(Animal):
  def __init__(self,name, type):
    self.type = type
    super().__init__(name) #super() is used to access parent constructor
    print("type", self.type)
    print("Dog constructor")

d = Dog("buddy","dog")
d.info()

```

```python
# polymorphism
# method overriding

class Animal:
    def sound(self):
        return "some generic sound"

class Dog(Animal):
    def sound(self):
        return "bark"

class Cat(Animal):
    def sound(self):
        return "meow"

obj = [Dog(),Cat(),Animal()]                   # runtime polymorphism
for i in obj:
    print(i.sound())
```

```python
# In polymorphism we can execute the same method in subclasses
class Vehicle:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model

    def move(self):
            pass

class Car(Vehicle):
    def move(self):
        print("drive")

class Boat(Vehicle):
    def move(self):
        print("sail")

class Plane(Vehicle):
    def move(self):
        print("fly")

car1 = Car("Ford","Mustang")
boat1 = Boat("Ibiza", "Touring 20")
plane1 = Plane("Boeing", "747")

for x in (car1,boat1,plane1):
    print(x.brand)
    print(x.model)
    x.move()
```

```python
# Abstraction
# Hiding implementation details and showing only important features to the user

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):    #Abstract method with no implementation
        pass

    def move(self):
        return "Moving"

class Dog(Animal):
    def make_sound(self):
        return "Bark"

dog = Dog()
print(dog.move())
```
