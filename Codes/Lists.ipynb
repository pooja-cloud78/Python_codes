```python
# Creating list
# emply list
L = []
print(L)

# 1D list
L = [3,4,5]
print(L)

# 2D list
L = [1,2,3,[4,5,6]]
print(L)

# Type conversion
L = list('Hello') # Converting string into list
print(L)
```

```python
# Accessing items from list
L = [1,2,3,4,5]
# Positive index
print(L[0])
print(L[3])
#print(L[9])        # out of range index

# Negative index
print(L[-2])
print(L[-4])

L = [1,2,3,[4,5],6]
print(L[3])      # Access 3rd element
print(L[3][0])  # Access element from list inside list

print(L[-1])
print(L[-2][-1])
print(L[-2][-2])
```

```python
# Slicing
L = [1,2,3,[4,5],6]
print(L[0:3])
print(L[3:])
print(L[:-3])
print(L[::2])
print(L[-5:-1:2])
print(L[::-1])   # reverse string
```

```python
# updating items in list
L = [1,2,3,4,5]
print(L)
L[-1] = 100
print(L)
L[1:4] = [200,300,400]   # Adding elements in the middle of list
print(L)
L[1:4] = [500]        # less elements than given range
print(L)
```

```python
# List oprations
L1 = [1, 2, 3]
L2 = [6,7]

print(L1+L2)
#print(L1*L2)   #can't multiply sequence by non-int of type 'list'
print(L2*2)

print(L1 == L2)
print(L1 != L2)
print(L1 > L2)
print(L1 < L2)

print(L1 and L2)
print(L1 or L2)

print(1 in L1)
print(6 not in L1)

# Looping through list
for i in L1:
  print(i)
```

```python
# Max, min, length of lists
L = [4, 5,6,7,True]
print(max(L))
print(min(L))
print(len(L))
print(sorted(L))
```

```python
# Adding element into list
L = [1,2,3,[4,5],6]

L.append(7)
L.append(True)
print(L)
L.append([8,9])  # Multiple elements add as another list
L.append('pune')
print(L)

# Extend
L.extend([0,4,False])   #Added separate multiple elements at once
print(L)
L.extend('Hello')         # for string it will
print(L)

# Insert
L = [1,2,3,4,5]
L.insert(0,100)
print(L)
```

```python
# removing elements from list
L = [1,2,3,4,5]
print(L)

#del L
del L[-1]      # Removes element based on index
print(L)
del L[1:3]        # Removes multiple elements
print(L)

# Remove
L = [1,2,3,4,5,3]
print(L)
L.remove(3)        # remove first occurrence of value
print(L)

# Clear
L.clear()
print(L)

# Pop
L = [1, 2, 3, 4, 5]
L.pop()                # Removing last element
L.pop(0)             # Removing based on index
print(L)
```

```python
# Sort method
thislist = ["banana", "Orange", "Kiwi", "cherry"]
# case sensitive sort
thislist.sort()
print(thislist)

thislist.sort(reverse = True)    # Descending
print(thislist)

# case insensitive sort
thislist.sort(key = str.lower)
print(thislist)
```

```python
# Reverse
fruits = ['banana', 'apple', 'cherry']
print(id(fruits[2]))
fruits.reverse()
print(fruits)

print(fruits[0])
print(id(fruits[0]))  # reverse will reverse order permanently
```

```python
# Count
L = [1,2,3,4,2,6,2,8,4]
print(L.count(2))
print(L.count(4))
```

```python
# List comprehension

# Add 1 to 10 numbers in list
L = []
for i in range(1,11):
  L.append(i)
print(L)

L = [ i for i in range(1,11)]
print(L)

v = [2,3,4]
s = -3
print(v*s)

# Add square
L = [1,2,3,4,5]
x = []
for i in L:
  x.append(i*i)
print(x)

x = [i*i for i in L]
print(x)
```

```python
# Find even numbers list using list comprehension

newlist = [x for x in range(11) if x % 2 == 0]
print(newlist)


# Cartesion product
L1 = [1,2,3,4]
L2 = [5,6,7,8]
c = []
for i in L1:
  for j in L2:
    c.append(i*j)
print(c)

c = [i*j for i in L1 for j in L2]
print(c)
```
